```python
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio

from heatpro.cold.standard_profile import StandardProfile

# Needed for correct layout of the produced html
pio.renderers.default = "notebook"
pd.options.plotting.backend = "plotly"
```

```python
print("Profiles availiable :")

for profile in StandardProfile:
    print(f"\t- {profile}")
```

## Working Day Profile

```python
working_day_profile = StandardProfile.WORKING_DAY.value

print(working_day_profile.description)
```

`StandardProfile.WORKING_DAY.value` and `StandardProfile.FULL_WEEK.value` are `WeekProfile` objects that have 3 attributes: `description`, `function` and `week_series_example`.

The easiest way to use these profile is by using `function` associated. 

These functions takes 2 arguments:

1. day of week (integer between 0 and 6)
2. hour of day (integer between 0 and 23)

```python
print(f"Profile weigth on tuesday 13 hour: {working_day_profile.function(1,13)}")
print(f"Profile weigth on sunday 16 hour: {working_day_profile.function(6,16)}")
```

You can easily apply the function from a `pandas.DatetimeIndex`.

```python
datetime_index = pd.date_range("2025-01-01","2025-01-05",periods=4)

print(f"datetime_index\n{datetime_index}")

week_profile_weigth = pd.DataFrame(
    {
        "day_of_week":datetime_index.day_of_week,
        "hour":datetime_index.hour,
    }
).apply(
    lambda row : working_day_profile.function(row.day_of_week,row.hour),
    axis=1
)

print(f"week_profile_weigth\n{week_profile_weigth}")
```

```python
go.Figure(
    go.Scatter(
        x=working_day_profile.week_series_example.index,
        y=working_day_profile.week_series_example,
        name="Working Day",
    ),
    layout_xaxis_tickformat="%a %H:%M",
    layout_xaxis_dtick=21600000 * 2,
    layout_title=f"{StandardProfile.WORKING_DAY} - Week Profile",
    layout_hovermode="x unified",
)
```
